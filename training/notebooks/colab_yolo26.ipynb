# Entrenamiento YOLO26n para Duckietown (Colab GPU)
Entrena, valida y exporta a ONNX para TensorRT en la Jetson Nano.

```python
!pip install ultralytics>=8.4.0 onnx onnxslim
```

```python
# Coloca el dataset en datasets/duckietown/{images,labels}/{train,val}
from ultralytics import YOLO
model = YOLO('yolo26n.pt')
model.train(data='datasets/duckietown.yaml', epochs=100, imgsz=416, device=0)
```

```python
# Exportar a ONNX (opset 12 para TensorRT 8.2)
YOLO('runs/detect/train/weights/best.pt').export(format='onnx', imgsz=416, opset=12, simplify=True)
```
